```python
from PIL import Image
import torch

from sam3.model_builder import build_sam3_image_model, _load_checkpoint
from sam3.model.sam3_image_processor import Sam3Processor

finetune_ckpt = r"C:\Users\keleslab\sam3\exp_runs\fly_ft\checkpoints\checkpoint.pt"
device = "cuda" if torch.cuda.is_available() else "cpu"

# 1) Load base SAM3 weights from HF
model = build_sam3_image_model(
    checkpoint_path=None,
    load_from_HF=True,
    device=device,
    eval_mode=True,
    enable_segmentation=True,
 )

# 2) Load your finetuned weights on top
_load_checkpoint(model, finetune_ckpt)

processor = Sam3Processor(model, device=device, confidence_threshold=0.05)

image = Image.open(r"C:\Users\keleslab\sam3\Test_Data\Samples\Sample1.png").convert("RGB")
state = processor.set_image(image)
state = processor.set_text_prompt(prompt="thorax", state=state)

masks = state["masks"]
boxes = state["boxes"]
scores = state["scores"]

import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(8, 8))
plt.imshow(image)

for m in masks.cpu().numpy():
    m = m.squeeze(0)  # (1, H, W) -> (H, W)
    plt.imshow(np.ma.masked_where(~m, m), alpha=0.3)

plt.axis("off")
plt.show()

plt.imshow(image)
```
